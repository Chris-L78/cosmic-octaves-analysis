## Section I: Nation Size Distribution Verification

Verifies that nations lack a characteristic physical length, unlike
biological organisms at each canonical scale. This supports the
argument that nations are supplementary (not canonical) structures
in the scale table.

```python
"""
Verify that nations lack characteristic physical length,
unlike biological organisms at each canonical scale.
"""
import math

# Nation sizes (approximate linear extent in km)
nations = {
    "Vatican City": 0.7,
    "Monaco": 2.0,
    "Singapore": 27,
    "Belgium": 280,
    "France": 950,
    "Brazil": 4300,
    "China": 5200,
    "Russia": 9000
}

# Biological organisms at one scale (bacteria, in micrometers)
bacteria = {
    "Mycoplasma": 0.3,
    "E. coli": 2.0,
    "B. subtilis": 4.0,
    "Thiomargarita": 750  # largest known, extreme outlier
}

# Calculate range in orders of magnitude
nation_sizes = list(nations.values())
bacteria_sizes = list(bacteria.values())

nation_range = math.log10(max(nation_sizes) / min(nation_sizes))
bacteria_range = math.log10(max(bacteria_sizes) / min(bacteria_sizes))
# Excluding Thiomargarita (extreme outlier):
bacteria_typical = [s for name, s in bacteria.items() if name != "Thiomargarita"]
bacteria_typical_range = math.log10(max(bacteria_typical) / min(bacteria_typical))

print("=== Nation Size Distribution ===")
for name, size in sorted(nations.items(), key=lambda x: x[1]):
    print(f"  {name}: {size} km")
print(f"  Range: {nation_range:.1f} orders of magnitude")
print()
print("=== Bacteria Size Distribution (micrometers) ===")
for name, size in sorted(bacteria.items(), key=lambda x: x[1]):
    print(f"  {name}: {size} um")
print(f"  Full range (with outlier): {bacteria_range:.1f} orders of magnitude")
print(f"  Typical range: {bacteria_typical_range:.1f} orders of magnitude")
print()
print("=== Conclusion ===")
print(f"Nations span {nation_range:.1f} orders of magnitude -- no characteristic length")
print(f"Typical bacteria span {bacteria_typical_range:.1f} orders of magnitude -- characteristic length ~1-5 um")
print(f"Ratio: nations are {nation_range/bacteria_typical_range:.0f}x more dispersed in size")
```

## Section II: Comprehensive Life-Test Scoring Verification

Programmatically reproduces the Individuality Gate (Michod's 5 Criteria)
and NASA 4 Requirements (Joyce 1994) assessments for all 15 structures
plus supplementary scales. Checks threshold calculations and validates
the substrate-independence conclusion.

```python
"""
Appendix C: Tests of Life -- Comprehensive Scoring Verification
================================================================
Verifies all Individuality Gate and NASA 4 Requirements scores
from the 15-scale assessment plus supplementary scales.

All scores are sourced from the detailed justifications in
agent-reports/individuality-gate-all-scales.md.
"""

# ------
# 1. DATA: Individuality Gate Scores (Michod's 5 Criteria)
# ------
# Order: Boundary, Coordination, Decision, Dissolution, Birth/Death

individuality_scores = {
    "Proton":               [3, 3, 1, 3, 0],
    "Atomic Orbital":       [3, 3, 1, 3, 0],
    "Ribosome":             [2, 3, 1, 2, 1],
    "Bacterium":            [3, 3, 3, 3, 3],
    "C. elegans":           [3, 3, 3, 3, 3],
    "Human":                [3, 3, 3, 3, 3],
    "City":                 [1, 3, 2, 2, 1],
    "Nation (supplementary)":[3, 3, 3, 3, 3],
    "Earth (rocky planet)": [3, 1, 0, 1, 0],
    "Sun":                  [3, 3, 2, 3, 3],
    "Solar System":         [3, 2, 1, 2, 1],
    "Local Bubble":         [2, 1, 0, 1, 1],
    "Open Cluster":         [2, 2, 1, 1, 2],
    "Milky Way":            [3, 3, 2, 3, 2],
    "Virgo Supercluster":   [1, 2, 1, 1, 1],
    "Observable Universe":  [0, 1, 0, 0, 0],
}

# ------
# 2. DATA: NASA 4 Requirements Scores (Joyce 1994)
# ------
# Order: Metabolism, Reproduction/Heredity, Homeostasis, Evolution

nasa_scores = {
    "Proton":               [0, 0, 1, 0],
    "Atomic Orbital":       [0, 0, 2, 0],
    "Ribosome":             [2, 1, 1, 1],
    "Bacterium":            [3, 3, 3, 3],
    "C. elegans":           [3, 3, 3, 3],
    "Human":                [3, 3, 3, 3],
    "City":                 [3, 2, 2, 1],
    "Nation (supplementary)":[3, 3, 3, 3],
    "Earth (rocky planet)": [2, 0, 2, 1],
    "Earth (Biosphere/Gaia)":[3, 1, 3, 2],
    "Sun":                  [3, 2, 3, 2],
    "Solar System":         [1, 1, 2, 1],
    "Local Bubble":         [3, 2, 2, 2],   # Updated per Ch 15: Zucker 2022 (star formation on shell) + Metha 2026 (chemical inheritance) raise score from 2 to 9
    "Open Cluster":         [1, 2, 1, 2],
    "Milky Way":            [3, 2, 3, 2],
    "Virgo Supercluster":   [1, 1, 1, 1],   # Updated per Ch 17: weak homeostasis still scores 1 (was 0). Total 4/12 Proto-living (was 3)
    "Observable Universe":  [2, 1, 1, 1],   # Updated per Ch 18 re-derivation: stellar fusion + structure formation + thermal inertia + cosmic web evolution = 5/12 Proto-living (was 0)
}

# ------
# 3. VERIFICATION: Totals and Classifications
# ------

def classify_individuality(scores):
    """Classify based on Individuality Gate total and threshold."""
    total = sum(scores)
    criteria_passing = sum(1 for s in scores if s >= 2)
    passes_gate = criteria_passing >= 4

    if total == 15:
        return total, "ORGANISM", passes_gate
    elif total >= 13:
        return total, "Borderline organism", passes_gate
    elif total >= 8:
        return total, "Organized system", passes_gate
    else:
        return total, "Aggregate", passes_gate


def classify_nasa(scores):
    """Classify based on NASA 4 Requirements total."""
    total = sum(scores)
    if total == 12:
        return total, "LIVING"
    elif total >= 9:
        return total, "Quasi-living"
    elif total >= 5:
        return total, "Proto-living"
    else:
        return total, "Non-living"


print("=" * 85)
print("INDIVIDUALITY GATE VERIFICATION (Michod's 5 Criteria)")
print("=" * 85)
print(f"{'Scale':<28} {'B':>3} {'C':>3} {'D':>3} {'R':>3} {'BD':>3} {'TOT':>5}  {'>=2':>3}  {'Classification'}")
print("-" * 85)

expected_totals_ind = {
    "Proton": 10, "Atomic Orbital": 10, "Ribosome": 9,
    "Bacterium": 15, "C. elegans": 15, "Human": 15,
    "City": 9, "Nation (supplementary)": 15,
    "Earth (rocky planet)": 5, "Sun": 14, "Solar System": 9,
    "Local Bubble": 5, "Open Cluster": 8, "Milky Way": 13,
    "Virgo Supercluster": 6, "Observable Universe": 1,
}

errors = []
for name, scores in individuality_scores.items():
    total, classification, passes = classify_individuality(scores)
    criteria_ge2 = sum(1 for s in scores if s >= 2)
    gate_str = "PASS" if passes else "FAIL"
    print(f"  {name:<26} {scores[0]:>3} {scores[1]:>3} {scores[2]:>3} "
          f"{scores[3]:>3} {scores[4]:>3} {total:>5}  {criteria_ge2}/5  {classification} ({gate_str})")

    if name in expected_totals_ind and total != expected_totals_ind[name]:
        errors.append(f"  MISMATCH: {name} expected {expected_totals_ind[name]}, got {total}")
```

```python
print()
print("=" * 85)
print("NASA 4 REQUIREMENTS VERIFICATION (Joyce 1994)")
print("=" * 85)
print(f"{'Scale':<28} {'Met':>4} {'Rep':>4} {'Hom':>4} {'Evo':>4} {'TOT':>5}  {'Classification'}")
print("-" * 85)

expected_totals_nasa = {
    "Proton": 1, "Atomic Orbital": 2, "Ribosome": 5,
    "Bacterium": 12, "C. elegans": 12, "Human": 12,
    "City": 8, "Nation (supplementary)": 12,
    "Earth (rocky planet)": 5, "Earth (Biosphere/Gaia)": 9,
    "Sun": 10, "Solar System": 5, "Local Bubble": 9,
    "Open Cluster": 6, "Milky Way": 10,
    "Virgo Supercluster": 4, "Observable Universe": 5,
}

for name, scores in nasa_scores.items():
    total, classification = classify_nasa(scores)
    print(f"  {name:<26} {scores[0]:>4} {scores[1]:>4} {scores[2]:>4} "
          f"{scores[3]:>4} {total:>5}  {classification}")

    if name in expected_totals_nasa and total != expected_totals_nasa[name]:
        errors.append(f"  MISMATCH: {name} expected {expected_totals_nasa[name]}, got {total}")

# ------
# 4. VERIFICATION: Substrate Independence Claim
# ------
print()
print("=" * 85)
print("SUBSTRATE INDEPENDENCE VERIFICATION")
print("=" * 85)

# The claim: organism/non-organism does NOT fall on biological/non-biological line
test_cases = [
    ("Biological ORGANISM",     "Bacterium",            True),
    ("Biological NON-organism", "Ribosome",             False),
    ("Social ORGANISM",         "Nation (supplementary)",True),
    ("Social NON-organism",     "City",                 False),
    ("Cosmic BORDERLINE",       "Sun",                  None),   # borderline
    ("Cosmic NON-organism",     "Earth (rocky planet)", False),
]

for label, name, expected_organism in test_cases:
    ind_total = sum(individuality_scores.get(name, [0]*5))
    ind_passes = sum(1 for s in individuality_scores.get(name, []) if s >= 2) >= 4
    nasa_total = sum(nasa_scores[name])

    if expected_organism is True:
        ok = ind_passes and nasa_total >= 9
    elif expected_organism is False:
        ok = not ind_passes or nasa_total < 9
    else:
        ok = True  # borderline -- just report

    status = "OK" if ok else "FAIL"
    print(f"  [{status}] {label:<30}  Ind: {ind_total}/15 ({'PASS' if ind_passes else 'FAIL'})  "
          f"NASA: {nasa_total}/12")
    if not ok:
        errors.append(f"  Substrate independence check failed for {label}")

# ------
# 5. VERIFICATION: Worked Example Totals
# ------
print()
print("=" * 85)
print("WORKED EXAMPLE VERIFICATION")
print("=" * 85)

worked_examples = {
    "Ant Colony (Strong)":  {"ind": [3,3,3,3,2], "nasa": [3,3,3,3], "passes": True},
    "Holobiont (Borderline)":{"ind": [2,2,2,1,1], "nasa": [3,1,2,1], "passes": False},
    "City (Fails)":         {"ind": [1,3,2,2,1], "nasa": [3,0,2,1], "passes": False},
}

for name, data in worked_examples.items():
    ind_total = sum(data["ind"])
    nasa_total = sum(data["nasa"])
    criteria_ge2 = sum(1 for s in data["ind"] if s >= 2)
    passes = criteria_ge2 >= 4

    status = "OK" if passes == data["passes"] else "FAIL"
    print(f"  [{status}] {name:<30}  Ind: {ind_total}/15 ({criteria_ge2}/5 >= 2)  "
          f"NASA: {nasa_total}/12  Gate: {'PASS' if passes else 'FAIL'}")

    if passes != data["passes"]:
        errors.append(f"  Worked example gate mismatch: {name}")

# ------
# 6. SUMMARY
# ------
print()
print("=" * 85)
if errors:
    print(f"VERIFICATION FAILED -- {len(errors)} error(s):")
    for e in errors:
        print(e)
else:
    print("ALL VERIFICATIONS PASSED")
    print(f"  - {len(individuality_scores)} Individuality Gate scores verified")
    print(f"  - {len(nasa_scores)} NASA 4 Requirements scores verified")
    print(f"  - {len(test_cases)} substrate independence checks passed")
    print(f"  - {len(worked_examples)} worked examples verified")
print("=" * 85)
```
