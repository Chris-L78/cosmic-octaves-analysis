## Section II — 15-Point Regression

Verifies the OLS regression on all 15 structures from the reference dataset.

```python
import numpy as np
from scipy import stats

data = [
    (8.4e-16,   1e-24,       "Proton"),
    (5.29e-11,  1.5e-16,     "Atom (H)"),
    (1.1e-8,    1e-3,        "Ribosome"),
    (1.0e-7,    6.0e2,       "Virus (T4)"),
    (1.0e-6,    1.2e3,       "Bacterium"),
    (5.0e-4,    2.5e-1,      "C. elegans"),
    (9.0e-1,    0.8,         "Human"),
    (6.37e6,    8.64e4,      "Earth"),
    (6.96e8,    3.16e7,      "Sun"),
    (3.0e17,    3.16e13,     "Open Cluster"),
    (9.5e17,    3.16e13,     "Giant Mol. Cloud"),
    (1.42e18,   4.7e13,      "Local Bubble"),
    (5.0e20,    2.26e15,     "Milky Way"),
    (1.5e24,    3.16e16,     "Supercluster"),
    (4.4e26,    4.35e17,     "Observable Universe"),
]

log_r = np.array([np.log10(r) for r, t, _ in data])
log_t = np.array([np.log10(t) for r, t, _ in data])

slope, intercept, r_value, p_value, std_err = stats.linregress(log_r, log_t)
n_pts = len(log_r)
dof = n_pts - 2
t_crit = stats.t.ppf(0.975, dof)

print("15-POINT REGRESSION")
print(f"n (slope):       {slope:.4f} +/- {std_err:.4f}")
print(f"intercept:       {intercept:.4f}")
print(f"95% CI for n:    [{slope - t_crit*std_err:.4f}, {slope + t_crit*std_err:.4f}]")
print(f"R-squared:       {r_value**2:.4f}")
print(f"Adjusted R-sq:   {1 - (1-r_value**2)*(n_pts-1)/dof:.4f}")
print(f"p-value:         {p_value:.6e}")
```

## Section V — Predictions & Residuals

Generates the prediction table comparing predicted vs. observed characteristic times
for all 15 structures and computes residual summary statistics.

```python
import numpy as np
from scipy import stats

data = [
    (8.4e-16,   1e-24,       "Proton"),
    (5.29e-11,  1.5e-16,     "Atom (H)"),
    (1.1e-8,    1e-3,        "Ribosome"),
    (1.0e-7,    6.0e2,       "Virus (T4)"),
    (1.0e-6,    1.2e3,       "Bacterium"),
    (5.0e-4,    2.5e-1,      "C. elegans"),
    (9.0e-1,    0.8,         "Human"),
    (6.37e6,    8.64e4,      "Earth"),
    (6.96e8,    3.16e7,      "Sun"),
    (3.0e17,    3.16e13,     "Open Cluster"),
    (9.5e17,    3.16e13,     "Giant Mol. Cloud"),
    (1.42e18,   4.7e13,      "Local Bubble"),
    (5.0e20,    2.26e15,     "Milky Way"),
    (1.5e24,    3.16e16,     "Supercluster"),
    (4.4e26,    4.35e17,     "Observable Universe"),
]

log_r = np.array([np.log10(r) for r, t, _ in data])
log_t = np.array([np.log10(t) for r, t, _ in data])

slope, intercept, r_value, p_value, std_err = stats.linregress(log_r, log_t)

print("PREDICTION TABLE")
print(f"{'Structure':<22} {'log10(T) obs':>13} {'log10(T) pred':>14} {'Residual':>9}")
print("-" * 62)
for r, t, name in sorted(data, key=lambda x: x[0]):
    lr = np.log10(r)
    lt_obs = np.log10(t)
    lt_pred = intercept + slope * lr
    resid = lt_obs - lt_pred
    print(f"{name:<22} {lt_obs:>13.2f} {lt_pred:>14.2f} {resid:>+9.2f}")

residuals = log_t - (intercept + slope * log_r)
print()
print(f"Mean |residual|: {np.mean(np.abs(residuals)):.2f}")
print(f"Median |residual|: {np.median(np.abs(residuals)):.2f}")
print(f"Std residual: {np.std(residuals, ddof=1):.4f}")
```

```python
import numpy as np
from scipy import stats

data = [
    (8.4e-16,   1e-24,       "Proton"),
    (5.29e-11,  1.5e-16,     "Atom (H)"),
    (1.1e-8,    1e-3,        "Ribosome"),
    (1.0e-7,    6.0e2,       "Virus (T4)"),
    (1.0e-6,    1.2e3,       "Bacterium"),
    (5.0e-4,    2.5e-1,      "C. elegans"),
    (9.0e-1,    0.8,         "Human"),
    (6.37e6,    8.64e4,      "Earth"),
    (6.96e8,    3.16e7,      "Sun"),
    (3.0e17,    3.16e13,     "Open Cluster"),
    (9.5e17,    3.16e13,     "Giant Mol. Cloud"),
    (1.42e18,   4.7e13,      "Local Bubble"),
    (5.0e20,    2.26e15,     "Milky Way"),
    (1.5e24,    3.16e16,     "Supercluster"),
    (4.4e26,    4.35e17,     "Observable Universe"),
]

log_r = np.array([np.log10(r) for r, t, _ in data])
log_t = np.array([np.log10(t) for r, t, _ in data])

slope, intercept, r_value, p_value, std_err = stats.linregress(log_r, log_t)

dof = len(log_r) - 2
t_crit = stats.t.ppf(0.975, dof)
ci_low = slope - t_crit * std_err
ci_high = slope + t_crit * std_err

z_070 = (slope - 0.70) / std_err
z_100 = (slope - 1.00) / std_err

print("TIME-RADIUS SCALING (Appendix F) \u2014 15-point dataset")
print(f"n (slope): {slope:.4f} +/- {std_err:.4f}")
print(f"intercept: {intercept:.4f}")
print(f"95% CI for n: [{ci_low:.4f}, {ci_high:.4f}]")
print(f"R-squared: {r_value**2:.4f}")
print(f"Adjusted R-squared: {1 - (1-r_value**2)*(len(log_r)-1)/dof:.4f}")
print(f"p-value: {p_value:.6e}")
print()
print(f"z vs n=0.70: {z_070:.2f}")
print(f"z vs n=1.00: {z_100:.2f}")
print()

# Prediction table
print("PREDICTION TABLE")
print(f"{'Structure':<22} {'log10(T) obs':>13} {'log10(T) pred':>14} {'Residual':>9}")
print("-" * 62)
for r, t, name in sorted(data, key=lambda x: x[0]):
    lr = np.log10(r)
    lt_obs = np.log10(t)
    lt_pred = intercept + slope * lr
    resid = lt_obs - lt_pred
    print(f"{name:<22} {lt_obs:>13.2f} {lt_pred:>14.2f} {resid:>+9.2f}")

residuals = log_t - (intercept + slope * log_r)
print()
print(f"Mean |residual|: {np.mean(np.abs(residuals)):.2f}")
print(f"Median |residual|: {np.median(np.abs(residuals)):.2f}")
print(f"Std residual: {np.std(residuals, ddof=1):.4f}")
```

## F.14 — Exponent Sensitivity

Tests how sensitive the crossover mass is to the choice of Kleiber exponent \u03b1.
Uses the mass-radius relation L = k \u00d7 M^\u03b3 with \u03b3 = 0.5, anchored to the
Open Cluster half-mass radius from Appendix A Table 1.

```python
import math

c        = 3.0e8       # m/s
tau_ref  = 3.0         # s (human subjective present)
M_ref    = 70.0        # kg
L_OC     = 4.70e16     # m  (half-mass radius; Appendix A Table 1)
M_OC     = 5.00e32     # kg (corrected mass;   Appendix A Table 1)
gamma    = 0.5
k = L_OC / (M_OC ** gamma)
print(f"k = L_OC / M_OC^0.5 = {k:.6f}")
print(f"Formula: M_cross = [k * 70^alpha / (3.0 * c)]^(1/(alpha - 0.5))\n")
print(f"{'alpha':>6} {'1/(alpha-0.5)':>14} {'M_cross (kg)':>16} {'log10(M_cross)':>15}")
print("-" * 56)

alpha_cases = [0.25, 0.23, 0.21, 0.20]
log_M_list  = []
for a in alpha_cases:
    inner = k * (M_ref ** a) / (tau_ref * c)
    exp   = 1.0 / (a - gamma)
    M_c   = inner ** exp
    lM    = math.log10(M_c)
    log_M_list.append(lM)
    print(f"{a:>6.2f} {exp:>14.3f} {M_c:>16.4e} {lM:>15.2f}")

delta_25_21 = log_M_list[0] - log_M_list[2]
print(f"\nCrossover shift alpha=0.25 -> 0.21: {delta_25_21:.2f} orders of magnitude")
print("(NOT ~1 order; root cause: alpha appears in both 70^alpha and 1/(alpha-0.5))")
```

## F.15 — Anchor Independence

Verifies that the crossover mass M_crossover is identical regardless of which
species is chosen as the metabolic anchor (Mouse, Human, or Elephant),
because the M_ref terms cancel algebraically.

```python
import math

c        = 3.0e8
tau_ref  = 3.0
M_ref    = 70.0
gamma    = 0.5
L_OC     = 4.70e16
M_OC     = 5.00e32
k = L_OC / (M_OC ** gamma)

ref_species = [("Mouse", 0.020), ("Human", 70.0), ("Elephant", 5000.0)]
print(f"{'Reference species':<16} {'M_ref (kg)':>12} {'tau_ref (s)':>12} {'M_cross (kg)':>16} {'log10(M)':>10}")
print("-" * 70)
for alpha_test in [0.25, 0.21]:
    print(f"\nalpha = {alpha_test}")
    for sp_name, M_r in ref_species:
        tau_r_adj = tau_ref * (M_r / 70.0) ** alpha_test
        inner_adj = k * (M_r ** alpha_test) / (tau_r_adj * c)
        exp_adj   = 1.0 / (alpha_test - gamma)
        M_c_adj   = inner_adj ** exp_adj
        print(f"  {sp_name:<14} {M_r:>12.3f} {tau_r_adj:>12.4f} {M_c_adj:>16.4e} {math.log10(M_c_adj):>10.2f}")
print()
print("M_crossover is identical for all reference species (M_ref cancels algebraically).")
```
