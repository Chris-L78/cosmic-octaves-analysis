# Appendix L: Golden-Ratio Harmonic Scaling -- Verification

This notebook reproduces all numerical claims in **Appendix L** of *Our Fractal Universe*:
Dirac large-number ratios from CODATA 2022 constants, harmonic indices,
deviations from the golden ratio, clustering probability, and the H0 prediction check.

## Setup: CODATA 2022 Constants and Golden Ratio

```python
import math

# -- CODATA 2022 fundamental constants --
k_e   = 8.9875517923e9       # Coulomb constant [N m^2 / C^2]
e_ch  = 1.602176634e-19      # elementary charge [C]
G     = 6.67430e-11          # gravitational constant [N m^2 / kg^2]
m_p   = 1.67262192369e-27    # proton mass [kg]
m_e   = 9.1093837015e-31     # electron mass [kg]
c     = 299792458.0           # speed of light [m/s]
r_e   = 2.8179403262e-15     # classical electron radius [m]
r_p   = 8.4075e-16           # proton charge radius [m]
k_B   = 1.380649e-23         # Boltzmann constant [J/K]

# Cosmological parameters
H0_planck = 67.4              # Planck 2018 [km/s/Mpc]
Mpc_in_m  = 3.0856775814913673e22  # 1 Megaparsec in metres
t_U_s     = 4.3509e17         # age of universe [seconds]

phi = (1.0 + math.sqrt(5)) / 2.0   # golden ratio

print("=" * 70)
print("APPENDIX L: GOLDEN-RATIO HARMONIC SCALING VERIFICATION")
print("=" * 70)
print(f"phi = {phi:.10f}")
```

## Ratio 1: F_EM / F_G (proton-electron)

```python
F_EM = k_e * e_ch**2
F_G  = G * m_p * m_e
R1   = F_EM / F_G
log_R1 = math.log10(R1)
h1 = log_R1 / 24.0

print(f"Ratio 1: F_EM / F_G (proton-electron)")
print(f"  F_EM = k_e * e^2  = {F_EM:.6e} N m^2")
print(f"  F_G  = G * mp * me = {F_G:.6e} N m^2")
print(f"  R1   = {R1:.6e}")
print(f"  log10(R1) = {log_R1:.4f}")
print(f"  h1 = {h1:.6f}   (appendix: 1.640)")
```

## Ratio 2: R_H / r_e (Hubble radius / electron radius)

```python
H0_SI = H0_planck * 1e3 / Mpc_in_m
R_H   = c / H0_SI
R2    = R_H / r_e
log_R2 = math.log10(R2)
h2 = log_R2 / 24.0

print(f"Ratio 2: R_H / r_e (Hubble radius / electron radius)")
print(f"  H0 = {H0_planck} km/s/Mpc = {H0_SI:.6e} s^-1")
print(f"  R_H = c / H0 = {R_H:.6e} m")
print(f"  R2  = R_H / r_e = {R2:.6e}")
print(f"  log10(R2) = {log_R2:.4f}")
print(f"  h2 = {h2:.6f}   (appendix: 1.695)")
```

## Ratio 3: t_U / (r_p / c)

```python
t_proton = r_p / c
R3 = t_U_s / t_proton
log_R3 = math.log10(R3)
h3 = log_R3 / 24.0

print(f"Ratio 3: t_U / (r_p / c)")
print(f"  r_p / c = {t_proton:.6e} s")
print(f"  R3 = {R3:.6e}")
print(f"  log10(R3) = {log_R3:.4f}")
print(f"  h3 = {h3:.6f}   (appendix: 1.716)")
```

## Summary: h-Index Deviations from phi and Probability

```python
print(f"{'='*70}")
print(f"SUMMARY: h-INDEX DEVIATIONS FROM phi = {phi:.6f}")
print(f"{'='*70}")
print(f"{'Ratio':<16s} {'h (computed)':>14s} {'h (appendix)':>14s} {'Dev from phi':>14s}")
print(f"{'-'*58}")
for name, hc, ha in [("F_EM/F_G", h1, 1.640),
                      ("R_H/r_e", h2, 1.695),
                      ("t_U/(r_p/c)", h3, 1.716)]:
    dev = (hc - phi) / phi * 100
    print(f"{name:<16s} {hc:>14.6f} {ha:>14.3f} {dev:>+13.2f}%")

# -- Probability calculation --
# If h ~ Uniform[1, 2], P(|h - phi| < 0.05) for one ratio:
window = 0.05
p_single = 2 * window / (2.0 - 1.0)  # = 0.10
p_three = p_single**3                 # = 0.001 = 0.1%

print(f"\nPROBABILITY OF CLUSTERING")
print(f"  Window: |h - phi| < {window}")
print(f"  P(single ratio): 2 * {window} / 1.0 = {p_single:.2f} = {p_single*100:.0f}%")
print(f"  P(all 3 ratios): {p_single}^3 = {p_three:.4f} = {p_three*100:.2f}%")
print(f"  i.e., roughly one in a thousand")
```

## H0 Prediction Check and L_n Formula

```python
# -- H0 prediction check --
exponent = 24 * phi
H0_pred_SI = c / (r_e * 10**exponent)
H0_pred_km = H0_pred_SI * Mpc_in_m / 1e3

print(f"H0 PREDICTION CHECK")
print(f"  24 * phi = {exponent:.4f}")
print(f"  H0 = c / (r_e * 10^(24*phi)) = {H0_pred_km:.1f} km/s/Mpc")
print(f"  Observed H0 (Planck): {H0_planck} km/s/Mpc")
print(f"  Ratio (predicted/observed): {H0_pred_km / H0_planck:.1f}x")
print(f"  NOTE: The exact CODATA log10(R_H/r_e) = {log_R2:.2f}, not 40.00")
print(f"        The ~70x discrepancy arises from dropping the coefficient")
print(f"        ~{10**(log_R2 - 40):.1f} when rounding R_H/r_e to '10^40'.")

# -- L_n formula check (phi^n vs phi*n) --
print(f"\nL_n FORMULA: phi^n vs phi*n")
for n in [1, 2, 3]:
    exp_times = 24 * phi * n
    exp_power = 24 * phi**n
    print(f"  n={n}: 24*phi*{n} = {exp_times:.2f}   24*phi^{n} = {exp_power:.2f}")
print(f"  n=2 with phi*n gives 10^{24*phi*2:.1f} (not 10^63)")
print(f"  n=2 with phi^n gives 10^{24*phi**2:.1f} (matches appendix 10^63)")
print(f"  CONCLUSION: Formula is L_n = 10^(24 * phi^n)")
```

## Uncertainty Propagation
